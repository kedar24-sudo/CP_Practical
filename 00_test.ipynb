```python
# get marks of a user and print the grade based on the marks
# above 90 - A , above 80 - B, above 70 - C, above 60 - D, below 60 - F
marks = float(input("Enter your marks: "))
if marks >= 90:  
    print("Grade: A")
elif marks >= 80:    
    print("Grade: B")
elif marks >= 70:
    print("Grade: C")
elif marks >= 60:
    print("Grade: D")
else:
    print("Grade: F")   
```

```python
# Activity 1
# Enter purity and classify the product.
# Write your solution here.
purity = float(input("Enter the purity of the product: "))
print(f"Purity: {purity}%")
if purity >= 99.0:
    classification = "Grade A"
elif purity >= 95.5:
    classification = "Grade B"
elif purity >= 90.0:
    classification = "Grade C"
else:
    classification = "Rejected"
print(f"The product is classified as: {classification}")  
```
