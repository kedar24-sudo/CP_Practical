Q1. In a Reactor with excess A the reaction A + B --> C reaches equillibrium initial concentraton are [a] = 2.0, [b] = 1.0 and [c] = 0 mol/l. if x mol/l of c is formed, kc = x/((2-x)(1-x)) which rearranges to the quadratic eq kc.x2 - (3kc + 1).x + 2kc = 0. 
write a python program that:
1.reads kc using input() prints the data type of the raw input and of the converted value, and shows a clear message no crash if the text is not a number or if kc<= 0
2. finds a,b,c the discriminant and both roots
3. selects the physically meaningful root (0 =< x =< 1, because b is limiting)
4. prints equilibrium [A],[B],[C](3 decimals) and the conversion of the limiting reactant B = x/[B]o X 100(%)
5. classifies the conversion using if,elif,else: => 70% "High", => 40% "moderate", otherwise "Low"

give proper python code on this covering all conditions without any error


```python
import math


def main():
    raw_kc = input("Enter the equilibrium constant Kc: ")
    print("Raw input type:", type(raw_kc))

    try:
        Kc = float(raw_kc)
    except (ValueError, OverflowError):
        print("Invalid input: enter Kc as a number.")
        return

    print("Converted Kc type:", type(Kc))

    if not math.isfinite(Kc) or Kc <= 0:
        print("Invalid Kc: it must be a positive, finite number.")
        return

    # Quadratic: Kc*x^2 - (3*Kc + 1)*x + 2*Kc = 0
    a = Kc
    b = -(3 * Kc + 1)
    c = 2 * Kc
    discriminant = b**2 - 4 * a * c

    print(f"\na = {a}")
    print(f"b = {b}")
    print(f"c = {c}")
    print(f"Discriminant = {discriminant}")

    if discriminant < 0:
        print("No real roots exist.")
        return

    root_1 = (-b - math.sqrt(discriminant)) / (2 * a)
    root_2 = (-b + math.sqrt(discriminant)) / (2 * a)
    print(f"Root 1 = {root_1:.6f}")
    print(f"Root 2 = {root_2:.6f}")

    # B is limiting, so product formation must satisfy 0 <= x <= [B]0.
    initial_A = 2.0
    initial_B = 1.0
    physical_roots = [root for root in (root_1, root_2)
                      if 0 <= root <= initial_B]

    if not physical_roots:
        print("No physically meaningful root was found.")
        return

    x = physical_roots[0]
    equilibrium_A = initial_A - x
    equilibrium_B = initial_B - x
    equilibrium_C = x
    conversion_B = (x / initial_B) * 100

    print(f"\nPhysical root x = {x:.3f} mol/L")
    print(f"[A] at equilibrium = {equilibrium_A:.3f} mol/L")
    print(f"[B] at equilibrium = {equilibrium_B:.3f} mol/L")
    print(f"[C] at equilibrium = {equilibrium_C:.3f} mol/L")
    print(f"B conversion = {conversion_B:.3f}%")

    if conversion_B >= 70:
        category = "High"
    elif conversion_B >= 40:
        category = "Moderate"
    else:
        category = "Low"

    print(f"Conversion category: {category}")


main()
```

Q2.
Part A Hydrogen Peroxide decomposes in a batch reactor by a first order reaction dC/dt=-kc. use the euler method: C_new = C_old - k.C_old.dt. Take C = 1.5 mol/l, K = 0.05 per min and dt = 2 min
using a for loop simulate 20 min an dprint a table of time (min) concentration(mol/l) and conversion(%)
using math.exp() calculate the exact value Ce^(-kt) at 20min and print the absolute error of the euler result
using a while loop repeat the simulation until conversion reaches 50% or time reaches 100min print the time needed and the final concentration
Part B. Reactor tags such as REACTOR-07/B-2026 are stores as text. Write a program that reads a tag loops through each character and prints the number of digits and the sum of the digits. show a message for an empty tag and for a tag with no digits. 
give a proper code covering all in python

```python
import math

# Part A: Euler simulation from 0 to 20 minutes.
initial_concentration = 1.5  # mol/L
rate_constant = 0.05         # per minute
time_step = 2.0              # minutes
simulation_time = 20.0       # minutes
number_of_steps = int(simulation_time / time_step)

concentration = initial_concentration
print("Time (min)\tConcentration (mol/L)\tConversion (%)")

for step in range(number_of_steps + 1):
    time = step * time_step
    conversion = (initial_concentration - concentration) / initial_concentration * 100
    print(f"{time:8.1f}\t{concentration:22.4f}\t{conversion:14.2f}")

    if step < number_of_steps:
        concentration = concentration - rate_constant * concentration * time_step

euler_concentration = concentration
exact_concentration = initial_concentration * math.exp(-rate_constant * simulation_time)
absolute_error = abs(euler_concentration - exact_concentration)

print(f"\nEuler concentration at {simulation_time:.0f} min: {euler_concentration:.6f} mol/L")
print(f"Exact concentration at {simulation_time:.0f} min: {exact_concentration:.6f} mol/L")
print(f"Absolute error: {absolute_error:.6f} mol/L")

# Repeat with a while loop until conversion reaches 50% or time reaches 100 min.
time = 0.0
concentration = initial_concentration
conversion = 0.0
maximum_time = 100.0

while conversion < 50.0 and time < maximum_time:
    step_size = min(time_step, maximum_time - time)
    concentration = concentration - rate_constant * concentration * step_size
    time += step_size
    conversion = (initial_concentration - concentration) / initial_concentration * 100

print(f"\nWhile-loop stopping time: {time:.1f} min")
print(f"Final concentration: {concentration:.4f} mol/L")
print(f"Final conversion: {conversion:.2f}%")

# Part B: Count and sum the digits in a reactor tag.
tag = input("Enter a reactor tag: ").strip()
digit_count = 0
sum_of_digits = 0

if not tag:
    print("The tag is empty.")
else:
    for character in tag:
        if character in "0123456789":
            digit_count += 1
            sum_of_digits += int(character)

    if digit_count == 0:
        print("The tag contains no digits.")

print(f"Number of digits: {digit_count}")
print(f"Sum of digits: {sum_of_digits}")
```

Q3. 
Part A. A lab has n candidate catalysts and wants to test r of them together (order does not matter). Write a recursive function factorial(n) (remember the base case for 0!) and a function combination(n,r)= n!/(r!(n-r)!) that uses it. Reject invalid inputs(negative, r>n). Print the number of possble catalysts sets for n = 7, r=2, and show the message for n=4,r=6
PartB. During a distillation run, samples are taken at times=Fibonacci number*5min. Write an itierative function fib_iter(n) that returns a list of the first n fibonacci numbers and a recursive function fib_rec(k) that returns the k^th fibonaccci number (Fo=1,F1=1). for n=9, check that both functions agree and print a table: sample no.|fibonaccci tag|sampling time min
write an python code on both the parts

```python
# Part A: Recursive factorial and catalyst combinations.
def factorial(n):
    if not isinstance(n, int) or isinstance(n, bool):
        raise ValueError("n must be a non-negative integer.")
    if n < 0:
        raise ValueError("n must be a non-negative integer.")
    if n == 0:
        return 1
    return n * factorial(n - 1)


def combination(n, r):
    if not isinstance(n, int) or isinstance(n, bool):
        raise ValueError("n must be a non-negative integer.")
    if not isinstance(r, int) or isinstance(r, bool):
        raise ValueError("r must be a non-negative integer.")
    if n < 0 or r < 0 or r > n:
        raise ValueError("Require n >= 0 and 0 <= r <= n.")
    return factorial(n) // (factorial(r) * factorial(n - r))


print("Part A: Catalyst combinations")
print(f"Number of ways to choose 2 catalysts from 7: {combination(7, 2)}")

try:
    combination(4, 6)
except ValueError as error:
    print(f"Invalid input for n=4, r=6: {error}")


# Part B: Fibonacci sequence with F0 = 1 and F1 = 1.
def fib_iter(n):
    if not isinstance(n, int) or isinstance(n, bool) or n < 0:
        raise ValueError("n must be a non-negative integer.")

    sequence = []
    first, second = 1, 1
    for _ in range(n):
        sequence.append(first)
        first, second = second, first + second
    return sequence


def fib_rec(k):
    if not isinstance(k, int) or isinstance(k, bool) or k < 0:
        raise ValueError("k must be a non-negative integer.")
    if k <= 1:
        return 1
    return fib_rec(k - 1) + fib_rec(k - 2)


number_of_samples = 9
fibonacci_tags = fib_iter(number_of_samples)
recursive_values = [fib_rec(index) for index in range(number_of_samples)]

print("\nPart B: Fibonacci sampling schedule")
if fibonacci_tags == recursive_values:
    print("Iterative and recursive Fibonacci results agree.")
else:
    raise AssertionError("The Fibonacci implementations do not agree.")

print(f"{'Sample No.':<12}{'Fibonacci Tag':<16}{'Sampling Time (min)'}")
for sample_number, fibonacci_tag in enumerate(fibonacci_tags, start=1):
    sampling_time = fibonacci_tag * 5
    print(f"{sample_number:<12}{fibonacci_tag:<16}{sampling_time}")
```

Q4.
Decomposition of hydrogen peroxide was studied with two catalyst loadings. the concentration (mol/l) was recorded:
at Time_min: 0,10,20,30,40,50
at Run1_mol_l(low loading): 1.20,0.98,0.80,0.66,0.54,0.44
at Run2_mol_l(high loading): 1.20,0.80,0.54,0.36,0.24,0.16
1.create a dataframe from this table. Save it as kinetics.csv without the index column and as kinetics.json . read the csv back and print its shape and data types
2.print the mean and standard deviation of Run1 and Run2 and state which run shows more variation
3.plot boths runs on one graph time on x-axis with title axis labels with units legend and grid
4. carbon dioxide is compressed slowly in a piston cylinder at constant temperature:
at Volume_L : 1,2,3,4,5
at pressure_bar : 3,02,1.49,1.01,0.76,0.59
plot pressure vs volume as a scatter plot calculate k as the mean of PxV and draw the reference curve P=k/v on the same graph add title labels with units and legend

give a python code on this covering all conditions

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Part 1: Create, save, and read back the kinetics data.
kinetics = pd.DataFrame({
    "Time_min": [0, 10, 20, 30, 40, 50],
    "Run1_mol_l": [1.20, 0.98, 0.80, 0.66, 0.54, 0.44],
    "Run2_mol_l": [1.20, 0.80, 0.54, 0.36, 0.24, 0.16],
})

kinetics.to_csv("kinetics.csv", index=False)
kinetics.to_json("kinetics.json", orient="records", indent=2)

kinetics_from_csv = pd.read_csv("kinetics.csv")
print("Kinetics data read from CSV:")
print(kinetics_from_csv)
print(f"\nDataFrame shape: {kinetics_from_csv.shape}")
print("Data types:")
print(kinetics_from_csv.dtypes)

# Part 2: Compare mean concentration and sample standard deviation.
run1_mean = kinetics_from_csv["Run1_mol_l"].mean()
run1_std = kinetics_from_csv["Run1_mol_l"].std()
run2_mean = kinetics_from_csv["Run2_mol_l"].mean()
run2_std = kinetics_from_csv["Run2_mol_l"].std()

print("\nSummary statistics:")
print(f"Run 1 mean: {run1_mean:.3f} mol/L; standard deviation: {run1_std:.3f} mol/L")
print(f"Run 2 mean: {run2_mean:.3f} mol/L; standard deviation: {run2_std:.3f} mol/L")
if run1_std > run2_std:
    print("Run 1 shows more variation by standard deviation.")
elif run2_std > run1_std:
    print("Run 2 shows more variation by standard deviation.")
else:
    print("Both runs have the same standard deviation.")

# Part 3: Plot both catalyst-loading runs.
plt.figure(figsize=(8, 5))
plt.plot(kinetics["Time_min"], kinetics["Run1_mol_l"], marker="o", label="Run 1 (low loading)")
plt.plot(kinetics["Time_min"], kinetics["Run2_mol_l"], marker="s", label="Run 2 (high loading)")
plt.title("Hydrogen Peroxide Decomposition")
plt.xlabel("Time (min)")
plt.ylabel("Concentration (mol/L)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

# Part 4: Pressure-volume data; the first pressure value is interpreted as 3.02 bar.
volume_L = [1, 2, 3, 4, 5]
pressure_bar = [3.02, 1.49, 1.01, 0.76, 0.59]
pressure_volume = pd.DataFrame({"Volume_L": volume_L, "Pressure_bar": pressure_bar})

k = (pressure_volume["Volume_L"] * pressure_volume["Pressure_bar"]).mean()
volume_curve = np.linspace(min(volume_L), max(volume_L), 100)
pressure_curve = k / volume_curve

plt.figure(figsize=(8, 5))
plt.scatter(volume_L, pressure_bar, color="tab:blue", label="Measured pressure")
plt.plot(volume_curve, pressure_curve, color="tab:orange", label=f"Reference curve (k = {k:.3f} bar L)")
plt.title("Carbon Dioxide Compression at Constant Temperature")
plt.xlabel("Volume (L)")
plt.ylabel("Pressure (bar)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
